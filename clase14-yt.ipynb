```python
#Vamos a escribir una funcion que haga todas las operaciones basicas sobre dos numeros
numero1= int(input("Introduce un numero"))
numero2= int(input("Introduce un segundo numero"))

#Funcion que realiza todas las operaciones basicas sobre dos numeros
def operaciones_basicas():
    #Realizar la suma de los numeros
    suma= numero1 + numero2
    resta= numero1 - numero2
    multiplicacion= numero1 * numero2
    division= numero1 / numero2

    #Imprimir por pantalla cada uno de los resultados
    print(f"La suma de los numeros es: {suma}.")
    print(f"La resta de los numeros es: {resta}")
    print(f"La multiplicacion de los numeros es: {multiplicacion}")
    print(f"La division de los numeros es: {division}")

operaciones_basicas()
```

```python
#Funcion que realiza todas las operaciones basicas sobre dos numeros, vamos a trabajar con parametros en esta ocasion
def operaciones_basicas(numero1, numero2):
    #Realizar la suma de los numeros
    suma= numero1 + numero2
    resta= numero1 - numero2
    multiplicacion= numero1 * numero2
    division= numero1 / numero2

    #Imprimir por pantalla cada uno de los resultados
    print(f"La suma de los numeros es: {suma}.")
    print(f"La resta de los numeros es: {resta}")
    print(f"La multiplicacion de los numeros es: {multiplicacion}")
    print(f"La division de los numeros es: {division}")
    print("-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-.-")

operaciones_basicas(5, 10)
operaciones_basicas(4, 2)
operaciones_basicas(-4,-10)
```
